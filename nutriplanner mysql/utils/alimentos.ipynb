```python
import pandas as pd
```

```python
alimentos = pd.read_csv("D:/faculdade/2_semestre/eng_software/NutriPlanner/tabela_TACO/alimentos.csv")
aminoacidos = pd.read_csv("D:/faculdade/2_semestre/eng_software/NutriPlanner/tabela_TACO/aminoacidos.csv")
acidosgraxos = pd.read_csv("D:/faculdade/2_semestre/eng_software/NutriPlanner/tabela_TACO/acidos-graxos.csv")
```

```python
df_merged = pd.merge(
    alimentos, 
    aminoacidos, 
    on=['Número do Alimento', 'Descrição dos alimentos'],
    how='left'
)

print(f"Shape após primeiro merge: {df_merged.shape}")
print(f"Colunas: {len(df_merged.columns)}")
```

```python
df_final = pd.merge(
    df_merged,
    acidosgraxos,
    on=['Número do Alimento', 'Descrição dos alimentos', 'Categoria do alimento'],
    how='left'
)

print(f"Shape final: {df_final.shape}")
print(f"Total de colunas: {len(df_final.columns)}")
```

```python
colunas = df_final.columns.tolist()
print(f"Total de colunas: {len(colunas)}")
print(f"Colunas únicas: {len(set(colunas))}")

duplicatas = [col for col in colunas if colunas.count(col) > 1]
if duplicatas:
    print(f"\nColunas duplicadas: {set(duplicatas)}")
```

```python
df_final = pd.merge(
    pd.merge(alimentos, aminoacidos, on=['Número do Alimento', 'Descrição dos alimentos'], how='left'),
    acidosgraxos,
    on=['Número do Alimento', 'Descrição dos alimentos', 'Categoria do alimento'],
    how='left'
)

def limpar_nome_coluna(nome):
    nome = nome.replace('....', ' (')
    nome = nome.replace('...', ' (')
    nome = nome.replace('..', ' (')
    
    if ' (' in nome and not nome.endswith(')'):
        nome = nome + ')'
    
    nome = nome.rstrip('.')
    
    return nome

df_final.columns = [limpar_nome_coluna(col) for col in df_final.columns]

print(df_final.columns.tolist())
```

```python
df = df_final.sort_values(by="Número do Alimento").reset_index(drop=True)
```

```python
df_final.fillna("-", inplace=True)
```
